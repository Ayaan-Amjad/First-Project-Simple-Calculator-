```python
# Week 1 Project: Simple Calculator
#This notebook takes two numbers and performs addition, subtraction, multiplication, and division.
```

```python
print("===========SIMPLE CALCULATOR============")

# Take two numbers from the variable
number1=float(input("Enter First number :"))
number2=float(input("Enter Second Number (Not Zero) :"))

#perform Calculatins
addition = number1 + number2
subtraction = number1 - number2
multiplication = number1 * number2
division = number1 / number2

#display the results
print("~~~~~RESULT~~~~~~")
print("Adddition :",addition)
print("Subtarction :",subtraction)
print("Multiplication :",multiplication)
print("Division :",division)
```
