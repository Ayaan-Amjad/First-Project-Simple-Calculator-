```python
%%writefile utility.py
import math
def factorial(number): 
    result = 1   
    for value in range(1, number + 1): 
        result = result * value   
    return result
def prime_check(number):
    if number < 2:
        return False  
    for value in range(2, number):
        if number % value == 0:  
            return False
    return True 
def square_root(number):
    return math.sqrt(number)
```

```python
import utility
```

```python
print("Factorial of 5:", utility.factorial(5))
print("Is 7 prime?:", utility.prime_check(7)) 
print("Square root of 25:", utility.square_root(25))
```

```python
print("===== UTILITY PROGRAM =====")
# Factorial
factorial_number = int(input("Enter a number for factorial: "))
if factorial_number < 0:  
    print("Factorial is not available for negative numbers.")
else:    
    answer = utility.factorial(factorial_number)  
    print("Factorial:", answer)
# Prime number check
prime_number = int(input("Enter a number to check prime: "))
if utility.prime_check(prime_number) == True: 
    print(prime_number, "is a prime number.") 
else:  
    print(prime_number, "is not a prime number.") 
# Square root
root_number = float(input("Enter a number for square root: "))
if root_number < 0:   
    print("Square root of a negative number is not allowed.")
else:   
    answer = utility.square_root(root_number)   
    print("Square root:", answer)
```
